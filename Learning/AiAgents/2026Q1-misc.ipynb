    Uvicorn 是 Python 的 ASGI 伺服器，負責把 HTTP 請求送進 FastAPI 應用程式。類比：

    ┌──────┬──────────────────────────┬──────────────────────────────┐
    │      │         Uvicorn          │            Nginx             │
    ├──────┼──────────────────────────┼──────────────────────────────┤
    │ 角色 │ 執行 Python app 的伺服器 │ 反向代理 / 靜態檔案伺服器    │
    ├──────┼──────────────────────────┼──────────────────────────────┤
    │ 語言 │ Python                   │ C                            │
    ├──────┼──────────────────────────┼──────────────────────────────┤
    │ 用途 │ 跑 FastAPI / Django 等   │ 轉發流量、負載均衡、SSL 終止 │
    └──────┴──────────────────────────┴──────────────────────────────┘

    實際生產環境通常是：Nginx → Uvicorn → FastAPI

    現在開發階段直接用 Uvicorn 就夠了，不需要 Nginx。
